### Classification Justification

Some FinTech companies can fit into more than one category, so three classifications required additional judgment:

1. Plaid — Payments & Digital Wallets: Plaid provides financial data connectivity that links users' bank accounts with financial applications. Although it is mainly an open-banking infrastructure provider, it supports digital financial transactions and payment-related services, so it is classified under Payments & Digital Wallets.

2. SoFi — Lending Tech & BNPL: SoFi provides several financial services, including lending, investing, and banking. Since lending is a major part of its FinTech business model, it is classified under Lending Tech & BNPL.

3. Chainalysis — Blockchain & Crypto: Chainalysis provides blockchain data, investigation, and compliance solutions. While its services also support regulatory compliance, its core technology is built around blockchain and cryptocurrency activity, so it is classified under Blockchain & Crypto.
